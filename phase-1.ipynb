## Numpy array and basics


```python
import numpy as np
```

### creating array from list

```python
arr_1d = np.array([1,2,3,4])
print(arr_1d)

arr_2d =  np.array([[1,2,3],[4,5,6]])
print(arr_2d)
```

### List vs numpy array


```python
py_list = [1,2,3,4,5]
print("python list multiplication",py_list*2) 

np_array = np.array([2,34,5,6]) # element wise multiplication
print("python array multiplication: ", np_array *2)
```

```python
import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n List operation time:",time.time() - start) 

start = time.time()
np_array = np.arange(1000000)*2 # arange is kitne range ka karana hai
print("\n Numpy operation time:",time.time() - start)
# numpy is efficient
```

### Creating array from scratch

```python
zeros = np.zeros((3,4)) 
print("zeros array:\n",zeros)

ones = np.ones((2,3))
print("ones array:\n",ones)

full = np.full((2,2),7)
print("full array:\n",full) 

random = np.random.rand(10)
print("random array (Normalization-0,1):\n",random) 

random2 = np.random.random((3,2))
print("Second random array:\n",random2)

random3 = np.random.randn(10)
print("Standardization(-3,3) Random array\n",random3)

randomint = np.random.randint(6)
print("random number between 0 & the given number:",randomint)

sequence = np.arange(0,10,2)
print("sequence array:\n",sequence) 

linear_space = np.linspace(0,1,5)
print("linear space array:\n",linear_space)
```

### Vector, Matrices and Tensor

```python
vector = np.array([1,2,3])
print("vector:\n",vector)

matrix = np.array([[1,2,3],
                   [4,5,6]])
print("matrix:\n",matrix)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]]) 
print("Tensor:\n",tensor)
```

### Array properties
Numpy array is optimized as it has only similar data types

```python
arr = np.array([[1,2,3],
                [4,5,6]]) 

print("Shape:\n",arr.shape) 
print("Dimension:\n",arr.ndim) 
print("Size:\n",arr.size) 
print("Dtype:\n",arr.dtype)  # this can be reduced as well. 

```

### Array Reshaping

```python
arr = np.arange(12)
print("original array:",arr)

reshaped = arr.reshape((3,4))
print("Reshaped array:\n",reshaped)

flattened = reshaped.flatten() # return copy
print("Flattened array:\n",flattened)

# ravel (return view, instead of copy) - this return original array
raveled = reshaped.ravel()
print("Raveled array:\n",raveled)

# Transpose 
transpose = reshaped.T
print("Transposed array:\n",transpose)
```
