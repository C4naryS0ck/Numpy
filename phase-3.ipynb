### Advanced Operation with Business examples

```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
# data structure : [ restaurant_id, 2021,2022,2023,2024]
sales_data = np.array ([
    [1,15000,18000,22000,25000], # Paradise Biryani
    [2,12000,14000,16000,19000], # Beijing Bites
    [3,20000,23000,26000,30000], # Pizza Hub
    [4,18000,21000,24000,27000], # Burger Point
    [5,16000,18000,20000,23000]  # Chai Point
])
```

```python
print("========== Zomato sales analysis =============")
print("\n Sales Data shape:",sales_data.shape)
print("\n Sample data for First three restaurant data:\n", sales_data[:3])   
print("\n Remove the first attribute and print the rest of data:\n", sales_data[:,1:])  
```

### Total Sales per year

```python
print(np.sum(sales_data, axis = 0))
print( np.sum(sales_data[:,1:],axis = 0)) 
```

### Minumum sales per restaurant.

```python
min_sales = np.min(sales_data[:,1:], axis = 1)
print(min_sales)
```

#### maximum sales per year

```python
max_sales = np.max(sales_data[:,1:], axis = 0) # column wise we will check!
print(max_sales)
```

#### Average sales per restaurant

```python
avg_sales = np.mean(sales_data[:,1:], axis = 1) 
print(avg_sales)
```

#### Cumulative Sales

```python
cumsum  = np.cumsum(sales_data[:,1:], axis = 1)
print(cumsum)
plt.figure(figsize = (8, 6)) 
plt.plot(np.mean(cumsum, axis=0)) 
plt.title("average Cumulative Sales per Year Across all restaurants:")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```

### Vector samples & operations

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])
print("Vector Addition: ", vector1 + vector2)
print("\nMultiplication vector", vector1 * vector2)
print("Dot product:",np.dot(vector1,vector2))
angle = np.arccos(np.dot(vector1,vector2)/(np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print(angle)
```

### Vectorized Operation

```python
restaurant_types = np.array(['Biryani',"chinese","pizza","burger","cafe"])

vectorized_upper = np.vectorize(str.upper)
print("Upper case letters:",vectorized_upper(restaurant_types))
```

### Broadcast: 

```python
monthly_average = sales_data[:, 1:] /  12 
# this divided by 12 is broadcasting to divide each number by 12.
print(monthly_average)

```
