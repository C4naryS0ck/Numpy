```python
s
```

```python
for i in range(0,9,3):  # iterates over row indices in steps of 3 -> 0,3,6.
    for j in range(0,9,3): # iterates over column indices in steps of 3 -> 0,3,6.
        n = s[i:i+3,j:j+3]
        block_sum = np.sum(n)
        print(block_sum)
```
